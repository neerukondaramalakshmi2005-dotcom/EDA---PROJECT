```python
import pandas as pd

# 1. Interpolate numerical columns (Works perfectly)
df['Age'] = df['Age'].interpolate()
print("--- After Age Interpolation ---")
print(df)
print("_________________________")

# 2. Fill categorical columns (Use ffill, bfill, or mode instead)
# 'ffill' (forward fill) propagates the last valid observation forward
df['Workclass'] = df['Workclass'].ffill()

# Fixed the typo here: 'Occupation' instead of 'Ocuupation'
df['Occupation'] = df['Occupation'].ffill()
print("--- After Workclass & Occupation Fill ---")
print(df)
print("_________________________")

# Alternatively, you can fill missing values with the most frequent value (mode)
most_frequent_country = df['Native Country'].mode()[0]
df['Native Country'] = df['Native Country'].fillna(most_frequent_country)

print("--- After Native Country Fill ---")
print(df)
```

```python
#  Detecting outlier ( it is )
# IQR ( Q1, Q2)
# LOWER_BOUND = Q1 - 1.5 * IQR
# UPPER_BOUND = Q2 + 1.5 * IQR
```

```python

# Detecting outlier
Q1 =df['Age'].quantile(0.25)
Q3 =df['Age'].quantile(0.75)
IQR = Q3 - Q1
LOWER_BOUND = Q1 - 1.5 * IQR
UPPER_BOUND = Q3 + 1.5 * IQR
print(LOWER_BOUND)
print(UPPER_BOUND)
```

```python
Q1 =df['Hours per Week'].quantile(0.25)
Q3 =df['Hours per Week'].quantile(0.75)
IQR = Q3 - Q1
LOWER_BOUND = Q1 - 1.5 * IQR
UPPER_BOUND = Q3 + 1.5 * IQR
print(LOWER_BOUND)
print(UPPER_BOUND)
```

```python
#EducationNum
Q1 =df['EducationNum'].quantile(0.75)
Q3 =df['EducationNum'].quantile(0.35)
IQR = Q3 - Q1
LOWER_BOUND = Q1 - 1.5 * IQR
UPPER_BOUND = Q3 + 1.5 * IQR
print(LOWER_BOUND)
print(UPPER_BOUND)
```

```python
# IDENTIFY IQR OUTLIER

iqr_outliers = df[(df['Age'] < LOWER_BOUND) | (df['Age'] > UPPER_BOUND)]
print("IQR Outerliers:", len(iqr_outliers))
```
